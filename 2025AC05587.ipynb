```python
import numpy as np
import pandas as pd
import matplotlib as plt
import seaborn
import streamlit
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score,matthews_corrcoef
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import MultinomialNB
from sklearn.ensemble import RandomForestClassifier
```

```python
df = pd.read_csv('Cancer_Data.csv')
```

```python
train_df, test_df = train_test_split(df, test_size=.2, random_state=42)
```

```python
df.head()
```

```python
print("shape:",df.shape)
print("size:",df.size)
print("datatypes:",df.dtypes)
```

```python
df.isna().any()
```

```python
df = df.drop(columns=["id"])
```

```python
X = df.drop(columns=['diagnosis'])
y = df['diagnosis']
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=101)
```

### Logistic Regression

```python

clf = LogisticRegression(max_iter=50000, random_state=53,penalty='l2', C=1.0, solver='saga')
clf.fit(X_train, y_train)

lr_acc = accuracy_score(y_test, clf.predict(X_test)) * 100
print(f"Logistic Regression model accuracy: {lr_acc:.2f}%")

lr_prec = precision_score(y_test, clf.predict(X_test), average='weighted') * 100
print(f"Logistic Regression model precision: {lr_prec:.2f}%")

lr_rec = recall_score(y_test, clf.predict(X_test), average='weighted') * 100
print(f"Logistic Regression model recall: {lr_rec:.2f}%")

lr_f1 = f1_score(y_test, clf.predict(X_test), average='weighted') * 100
print(f"Logistic Regression model F1-score: {lr_f1:.2f}%")

lr_mcc = matthews_corrcoef(y_test, clf.predict(X_test))
print(f"Logistic Regression model Matthews correlation coefficient: {lr_mcc:.2f}")
```

### Decision Tree Classifier

```python
dtc = DecisionTreeClassifier(random_state=1,criterion='entropy',max_depth=6,splitter='best',min_samples_split=2,min_samples_leaf=1)
dtc.fit(X_train, y_train)

dtc_acc = accuracy_score(y_test, dtc.predict(X_test)) * 100
print(f"Decision Tree model accuracy: {dtc_acc:.2f}%")

dtc_prec = precision_score(y_test, dtc.predict(X_test), average='weighted') * 100
print(f"Decision Tree model precision: {dtc_prec:.2f}%")

dtc_rec = recall_score(y_test, dtc.predict(X_test), average='weighted') * 100
print(f"Decision Tree model recall: {dtc_rec:.2f}%")

dtc_f1 = f1_score(y_test, dtc.predict(X_test), average='weighted') * 100
print(f"Decision Tree model F1-score: {dtc_f1:.2f}%")

dtc_mcc = matthews_corrcoef(y_test, dtc.predict(X_test))
print(f"Decision Tree model Matthews correlation coefficient: {dtc_mcc:.2f}")
```

```python

knnc = KNeighborsClassifier(n_neighbors=9, weights='uniform', algorithm='auto', leaf_size=20, p=2, metric='minkowski', metric_params=None, n_jobs=None)
knnc.fit(X_train, y_train)

knnc_acc = accuracy_score(y_test, knnc.predict(X_test)) * 100
print(f"K-Nearest Neighbors model accuracy: {knnc_acc:.2f}%")

knnc_prec = precision_score(y_test, knnc.predict(X_test), average='weighted') * 100
print(f"K-Nearest Neighbors model precision: {knnc_prec:.2f}%")

knnc_rec = recall_score(y_test, knnc.predict(X_test), average='weighted') * 100
print(f"K-Nearest Neighbors model recall: {knnc_rec:.2f}%")

knnc_f1 = f1_score(y_test, knnc.predict(X_test), average='weighted') * 100
print(f"K-Nearest Neighbors model F1-score: {knnc_f1:.2f}%")

knnc_mcc = matthews_corrcoef(y_test, knnc.predict(X_test))
print(f"K-Nearest Neighbors model Matthews correlation coefficient: {knnc_mcc:.2f}")
```

```python

nbc = MultinomialNB(alpha=1.0,force_alpha=True, fit_prior=True, class_prior=None)
nbc.fit(X_train, y_train)

nbc_acc = accuracy_score(y_test, nbc.predict(X_test)) * 100
print(f"Multinomial NB model accuracy: {nbc_acc:.2f}%")

nbc_prec = precision_score(y_test, nbc.predict(X_test), average='weighted') * 100
print(f"Multinomial NB model precision: {nbc_prec:.2f}%")

nbc_rec = recall_score(y_test, nbc.predict(X_test), average='weighted') * 100
print(f"Multinomial NB model recall: {nbc_rec:.2f}%")

nbc_f1 = f1_score(y_test, nbc.predict(X_test), average='weighted') * 100
print(f"Multinomial NB model F1-score: {nbc_f1:.2f}%")

nbc_mcc = matthews_corrcoef(y_test, nbc.predict(X_test))
print(f"Multinomial NB model Matthews correlation coefficient: {nbc_mcc:.2f}")
```

```python
rf = RandomForestClassifier(
            n_estimators=300,       # Number of trees
            max_depth=20,         # No max depth limit
            random_state=42,        # Reproducibility
            n_jobs=2              # Use all CPU cores
        )
rf.fit(X_train, y_train)

rf_acc = accuracy_score(y_test, rf.predict(X_test)) * 100
print(f"Random Forest model accuracy: {rf_acc:.2f}%")

rf_prec = precision_score(y_test, rf.predict(X_test), average='weighted') * 100
print(f"Random Forest model precision: {rf_prec:.2f}%")

rf_rec = recall_score(y_test, rf.predict(X_test), average='weighted') * 100
print(f"Random Forest model recall: {rf_rec:.2f}%")

rf_f1 = f1_score(y_test, rf.predict(X_test), average='weighted') * 100
print(f"Random Forest model F1-score: {rf_f1:.2f}%")

rf_mcc = matthews_corrcoef(y_test, rf.predict(X_test))
print(f"Random Forest model Matthews correlation coefficient: {rf_mcc:.2f}")
```
